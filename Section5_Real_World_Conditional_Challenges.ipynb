## Q1: ATM Withdrawal - Advanced Challenge

PIN correct, account active, amount positive, multiple of Rs.100, balance sufficient, Rs.500 minimum balance must remain, amount > Rs.20,000 -> extra verification.

```python
correct_pin = 1234
entered_pin = 1234
account_active = True
balance = 30000
amount = 21000
extra_verified = False   # amount > 20000 ho to extra verification chahiye

if entered_pin == correct_pin:
    if account_active:
        if amount > 0:
            if amount % 100 == 0:
                if balance >= amount:
                    if balance - amount >= 500:
                        if amount > 20000 and not extra_verified:
                            print("Extra verification required")
                        else:
                            balance -= amount
                            print("Withdrawal successful. Balance:", balance)
                    else:
                        print("Minimum balance of Rs.500 must remain")
                else:
                    print("Insufficient Balance")
            else:
                print("Amount must be a multiple of 100")
        else:
            print("Amount must be positive")
    else:
        print("Account is inactive")
else:
    print("Invalid PIN")
```

## Q2: Electricity Bill

0-100 units Rs.5, 101-200 Rs.7, 201-300 Rs.10, above 300 Rs.15 per unit. Bill > Rs.3,000 -> 10% surcharge.

```python
units = 250

# Slab-wise billing (har slab ke units apne rate par)
if units <= 100:
    bill = units * 5
elif units <= 200:
    bill = 100 * 5 + (units - 100) * 7
elif units <= 300:
    bill = 100 * 5 + 100 * 7 + (units - 200) * 10
else:
    bill = 100 * 5 + 100 * 7 + 100 * 10 + (units - 300) * 15

if bill > 3000:
    bill = bill + bill * 0.10   # 10% surcharge

print("Units:", units)
print("Total Bill: Rs.", bill)
```

## Q3: Employee Bonus

Experience >= 1 and salary >= 50,000: performance >= 90 -> 20%, >= 75 -> 10%, else 5%. Salary < 50,000: performance >= 90 -> 15%, else 5%.

```python
experience = 2        # years
salary = 60000
performance = 80
salary_for_bonus = salary

if experience >= 1 and salary >= 50000:
    if performance >= 90:
        bonus_percent = 20
    elif performance >= 75:
        bonus_percent = 10
    else:
        bonus_percent = 5
elif salary < 50000:
    if performance >= 90:
        bonus_percent = 15
    else:
        bonus_percent = 5
else:
    # experience < 1 aur salary >= 50000 (question mein diya nahi) -> 5% maan liya
    bonus_percent = 5

bonus_amount = salary * bonus_percent / 100
print("Bonus %:", bonus_percent)
print("Bonus Amount: Rs.", bonus_amount)
```
