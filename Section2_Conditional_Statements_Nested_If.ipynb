## Q1: ATM Withdrawal

Conditions:
- Check whether the PIN is correct.
- If correct, check whether amount is a multiple of 100.
- If yes, check whether balance is sufficient.
- If sufficient, withdraw. Otherwise print "Insufficient Balance".
- If PIN is incorrect, print "Invalid PIN".

```python
correct_pin = 1234
entered_pin = 1234
balance = 10000
amount = 2500

if entered_pin == correct_pin:
    if amount % 100 == 0:
        if balance >= amount:
            balance -= amount
            print("Withdrawal successful. Remaining balance:", balance)
        else:
            print("Insufficient Balance")
    else:
        print("Amount must be a multiple of 100")
else:
    print("Invalid PIN")
```

## Q2: Student Result System

Marks of 3 subjects. Any subject < 40 -> Fail.
Otherwise percentage: >=75 Distinction, >=60 First Division, >=50 Second Division, else Pass.

```python
m1, m2, m3 = 70, 65, 80

if m1 >= 40 and m2 >= 40 and m3 >= 40:
    percentage = (m1 + m2 + m3) / 3
    print("Percentage:", round(percentage, 2))
    if percentage >= 75:
        print("Distinction")
    elif percentage >= 60:
        print("First Division")
    elif percentage >= 50:
        print("Second Division")
    else:
        print("Pass")
else:
    print("Fail")
```

## Q7: Electricity Bill

0-100 units Rs.5, 101-200 Rs.7, 201-300 Rs.10, above 300 Rs.15 per unit. Bill > Rs.3,000 -> 10% surcharge.

(Yahan slab-wise billing use ki hai.)

```python
units = 250

# Slab-wise billing (har slab ke units apne rate par)
if units <= 100:
    bill = units * 5
elif units <= 200:
    bill = 100 * 5 + (units - 100) * 7
elif units <= 300:
    bill = 100 * 5 + 100 * 7 + (units - 200) * 10
else:
    bill = 100 * 5 + 100 * 7 + 100 * 10 + (units - 300) * 15

if bill > 3000:
    bill = bill + bill * 0.10   # 10% surcharge

print("Units:", units)
print("Total Bill: Rs.", bill)
```

## Q10: ATM - Difficult Challenge

PIN correct, account active, amount positive, multiple of 100, balance sufficient, Rs.500 minimum balance after withdrawal, amount > Rs.20,000 -> extra verification.

```python
correct_pin = 1234
entered_pin = 1234
account_active = True
balance = 30000
amount = 21000
extra_verified = False   # amount > 20000 ho to extra verification chahiye

if entered_pin == correct_pin:
    if account_active:
        if amount > 0:
            if amount % 100 == 0:
                if balance >= amount:
                    if balance - amount >= 500:
                        if amount > 20000 and not extra_verified:
                            print("Extra verification required")
                        else:
                            balance -= amount
                            print("Withdrawal successful. Balance:", balance)
                    else:
                        print("Minimum balance of Rs.500 must remain")
                else:
                    print("Insufficient Balance")
            else:
                print("Amount must be a multiple of 100")
        else:
            print("Amount must be positive")
    else:
        print("Account is inactive")
else:
    print("Invalid PIN")
```
