# Section 4: List Comprehensions (17 Questions)


## Question 1

Find numbers greater than 50.

```python
numbers = [20, 55, 30, 80, 45, 90, 10]
result = [n for n in numbers if n > 50]
print(result)
```

## Question 2

Find numbers between 20 and 50.

```python
numbers = [10, 25, 35, 60, 45, 15, 50, 70]
result = [n for n in numbers if 20 < n < 50]
print(result)
```

## Question 3

Find numbers divisible by both 3 and 5.

```python
numbers = [10, 15, 20, 30, 45, 50, 60, 75]
result = [n for n in numbers if n % 3 == 0 and n % 5 == 0]
print(result)
```

## Question 4

Find names having more than 4 characters.

```python
names = ["Amit", "Rahul", "Neha", "Rohit", "Sneha"]
result = [n for n in names if len(n) > 4]
print(result)
```

## Question 5

Find names starting with "A".

```python
names = ["Amit", "Rahul", "Ankit", "Neha", "Arjun", "Rohit"]
result = [n for n in names if n.startswith("A")]
print(result)
```

## Question 6

Extract vowels from a string.

```python
text = "programming"
result = [ch for ch in text if ch in "aeiou"]
print(result)
```

## Question 7

Extract digits from a string.

```python
data = "Python123Programming456"
result = [ch for ch in data if ch.isdigit()]
print(result)
```

## Question 8

Replace even with "Even", odd with "Odd".

Expected: `['Even', 'Odd', 'Even', 'Odd', 'Even']`

```python
numbers = [10, 15, 20, 25, 30]
result = ["Even" if n % 2 == 0 else "Odd" for n in numbers]
print(result)
```

## Question 9

Replace negative numbers with 0.

Expected: `[10, 0, 20, 0, 30, 0]`

```python
numbers = [10, -5, 20, -8, 30, -2]
result = [n if n >= 0 else 0 for n in numbers]
print(result)
```

## Question 10

10% discount if price >= 1000, else 5%.

Expected: `[475.0, 1080.0, 760.0, 1800.0, 1350.0]`

```python
prices = [500, 1200, 800, 2000, 1500]
result = [p * 0.90 if p >= 1000 else p * 0.95 for p in prices]
print(result)
```

## Question 11

Employees earning more than Rs.50,000.

Expected: `['Rahul', 'Neha']`

```python
employees = [
    {"name": "Amit", "salary": 45000},
    {"name": "Rahul", "salary": 60000},
    {"name": "Neha", "salary": 75000},
    {"name": "Rohit", "salary": 35000}
]
result = [e["name"] for e in employees if e["salary"] > 50000]
print(result)
```

## Question 12

Products costing more than Rs.1,000.

Expected: `[1299, 2499, 1599]`

```python
prices = [499, 1299, 799, 2499, 999, 1599]
result = [p for p in prices if p > 1000]
print(result)
```

## Question 13

20% discount only to products costing more than Rs.2,000.

Expected: `[2000.0, 2400.0, 4000.0]`

```python
prices = [500, 1200, 2500, 3000, 1500, 5000]
result = [p * 0.80 for p in prices if p > 2000]
print(result)
```

## Question 14: Bank Transactions

Find all deposits and all withdrawals.

```python
transactions = [5000, -2000, 3000, -1000, 7000, -500]
deposits = [t for t in transactions if t > 0]
withdrawals = [t for t in transactions if t < 0]
print("Deposits:", deposits)
print("Withdrawals:", withdrawals)
```

## Question 15

Names of students who scored 60 or more.

Expected: `['Amit', 'Neha']`

```python
students = [
    {"name": "Amit", "marks": 75},
    {"name": "Rahul", "marks": 45},
    {"name": "Neha", "marks": 90},
    {"name": "Rohit", "marks": 55}
]
result = [s["name"] for s in students if s["marks"] >= 60]
print(result)
```

## Question 16

Names of employees from the IT department.

Expected: `['Amit', 'Neha']`

```python
employees = [
    {"name": "Amit", "department": "IT"},
    {"name": "Rahul", "department": "HR"},
    {"name": "Neha", "department": "IT"},
    {"name": "Rohit", "department": "Sales"}
]
result = [e["name"] for e in employees if e["department"] == "IT"]
print(result)
```

## Question 17

Employees who earn more than Rs.60,000 and belong to IT.

Expected: `['Neha']`

```python
employees = [
    {"name": "Amit", "department": "IT", "salary": 55000},
    {"name": "Rahul", "department": "HR", "salary": 70000},
    {"name": "Neha", "department": "IT", "salary": 80000},
    {"name": "Rohit", "department": "IT", "salary": 45000}
]
result = [e["name"] for e in employees if e["department"] == "IT" and e["salary"] > 60000]
print(result)
```
