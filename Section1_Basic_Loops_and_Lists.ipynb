## Q10: Count even and odd (1 to 50)

Count even and odd numbers between 1 and 50.

```python
even_count = 0
odd_count = 0
for i in range(1, 51):
    if i % 2 == 0:
        even_count += 1
    else:
        odd_count += 1
print("Even:", even_count)
print("Odd:", odd_count)
```

## Q11: Reverse a list

Print the elements of a list in reverse order.

`data = [22, 33, 12, 32, 45, 13]`

```python
data = [22, 33, 12, 32, 45, 13]
for item in reversed(data):
    print(item)
```

## Q12: Largest number

Find the largest number in a list using a for loop.

`data = [22, 55, 12, 89, 34, 67]`

```python
data = [22, 55, 12, 89, 34, 67]
largest = data[0]
for n in data:
    if n > largest:
        largest = n
print("Largest:", largest)
```

## Q13: Smallest number

Find the smallest number in a list.

`data = [22, 55, 12, 89, 34, 67]`

```python
data = [22, 55, 12, 89, 34, 67]
smallest = data[0]
for n in data:
    if n < smallest:
        smallest = n
print("Smallest:", smallest)
```

## Q14: Count vowels

Count vowels in a string.

```python
text = "Programming Is Fun"
count = 0
for ch in text.lower():
    if ch in "aeiou":
        count += 1
print("Vowels:", count)
```

## Q15: Count spaces

Count the number of spaces in a sentence.

`text = "Python is easy to learn"`

```python
text = "Python is easy to learn"
spaces = 0
for ch in text:
    if ch == " ":
        spaces += 1
print("Spaces:", spaces)
```
