## Question 1: Check Student Result

Marks >= 40 -> Pass, else Fail.

````python
students = {"Rahul": 75, "Amit": 32, "Priya": 88, "Neha": 39, "Rohit": 55}

for name, marks in students.items():
    if marks >= 40:
        print(name, "Pass")
    else:
        print(name, "Fail")
````

## Question 2: Student Performance (Nested If)

Marks >= 80 Excellent, >= 60 Good, >= 40 Average, < 40 Fail.

````python
students = {"Rahul": 85, "Amit": 35, "Priya": 72, "Neha": 48}

for name, marks in students.items():
    if marks >= 40:
        if marks >= 80:
            print(name, "Excellent")
        elif marks >= 60:
            print(name, "Good")
        else:
            print(name, "Average")
    else:
        print(name, "Fail")
````

## Question 3: Product Stock Management

Stock == 0 Out of Stock, <= 5 Low Stock, > 5 Available.

````python
products = {"Laptop": 10, "Mouse": 3, "Keyboard": 0, "Monitor": 7, "Printer": 2}

for product, stock in products.items():
    if stock == 0:
        print(product, "- Out of Stock")
    else:
        if stock <= 5:
            print(product, "- Low Stock")
        else:
            print(product, "- Available")
````

## Question 4: Employee Salary and Department

Salary >= 50,000: IT -> Senior IT Employee, HR -> Senior HR Employee, others -> Senior Employee. Salary < 50,000 -> Junior Employee.

````python
employees = {
    "Rahul": {"salary": 60000, "department": "IT"},
    "Amit": {"salary": 40000, "department": "HR"},
    "Priya": {"salary": 70000, "department": "IT"},
    "Neha": {"salary": 45000, "department": "Finance"}
}

for name, info in employees.items():
    if info["salary"] >= 50000:
        if info["department"] == "IT":
            print(name, "- Senior IT Employee")
        elif info["department"] == "HR":
            print(name, "- Senior HR Employee")
        else:
            print(name, "- Senior Employee")
    else:
        print(name, "- Junior Employee")
````

## Question 5: Check Even/Odd Values

Even and > 20 -> Large Even, Even and <= 20 -> Even, Odd -> Odd.

````python
data = {"a": 10, "b": 15, "c": 25, "d": 30, "e": 40}

for key, value in data.items():
    if value % 2 == 0:
        if value > 20:
            print(key, value, "Large Even")
        else:
            print(key, value, "Even")
    else:
        print(key, value, "Odd")
````

## Question 6: Shopping Cart Classification

Price >= 5000 Premium, 1000 to < 5000 Expensive, < 1000 Affordable.

````python
products = {"Mobile": 15000, "Mouse": 500, "Laptop": 60000, "Keyboard": 1200}

for product, price in products.items():
    if price >= 5000:
        print(product, "- Premium Product")
    elif price >= 1000:
        print(product, "- Expensive Product")
    else:
        print(product, "- Affordable Product")
````

## Question 7: Student Eligibility

Attendance >= 75 and marks >= 40 -> Eligible; attendance >= 75 and marks < 40 -> Failed in Exam; attendance < 75 -> Not Eligible.

````python
students = {
    "Rahul": {"attendance": 85, "marks": 70},
    "Amit": {"attendance": 60, "marks": 80},
    "Priya": {"attendance": 90, "marks": 35},
    "Neha": {"attendance": 78, "marks": 55}
}

for name, info in students.items():
    if info["attendance"] >= 75:
        if info["marks"] >= 40:
            print(name, "- Eligible")
        else:
            print(name, "- Failed in Exam")
    else:
        print(name, "- Not Eligible")
````

## Question 8: Bank Account Classification

Balance >= 10,000 Premium, > 0 Regular, == 0 No Balance, < 0 Overdraft.

````python
accounts = {"Rahul": 25000, "Amit": 5000, "Priya": 0, "Neha": -2000}

for name, balance in accounts.items():
    if balance >= 10000:
        print(name, "- Premium Account")
    elif balance > 0:
        print(name, "- Regular Account")
    elif balance == 0:
        print(name, "- No Balance")
    else:
        print(name, "- Overdraft")
````

## Question 9: Nested Dictionary - Employee Experience

## Question 10: Student Marks and Attendance

Attendance < 75 -> Not Eligible. Otherwise: marks >= 80 Distinction, >= 60 First Division, >= 40 Pass, else Fail.

Expected Output:
```
Rahul Distinction
Amit Fail
Priya Not Eligible
Neha Pass
Rohit Not Eligible
```

````python
students = {
    "Rahul": {"marks": 85, "attendance": 90},
    "Amit": {"marks": 35, "attendance": 80},
    "Priya": {"marks": 65, "attendance": 70},
    "Neha": {"marks": 45, "attendance": 85},
    "Rohit": {"marks": 90, "attendance": 60}
}

for name, info in students.items():
    if info["attendance"] < 75:
        print(name, "Not Eligible")
    else:
        if info["marks"] >= 80:
            print(name, "Distinction")
        elif info["marks"] >= 60:
            print(name, "First Division")
        elif info["marks"] >= 40:
            print(name, "Pass")
        else:
            print(name, "Fail")
````
